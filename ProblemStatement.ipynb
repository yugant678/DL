```python
def answer_with_sources(question):

    docs = db.similarity_search(
        question,
        k=3
    )

    context = "\n\n".join(
        [doc.page_content for doc in docs]
    )

    response = llm.invoke(
        f"""
Answer only using the context.

Context:
{context}

Question:
{question}
"""
    )

    sources = []

    for doc in docs:
        src = doc.metadata.get(
            "source",
            "Unknown"
        )

        if src not in sources:
            sources.append(src)

    return {
        "answer": response.content,
        "sources": sources
    }
```

```python
import streamlit as st

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_ollama import ChatOllama

import tempfile
import os
import re

# --------------------------
# CONFIG
# --------------------------

CHROMA_DIR = "chroma_db"

# --------------------------
# CLEAN TEXT
# --------------------------

def clean_text(text):
    text = re.sub(r"<pad>", "", text)
    text = re.sub(r"<EOS>", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

# --------------------------
# LOAD EMBEDDING MODEL
# --------------------------

@st.cache_resource
def load_embeddings():
    return HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )

# --------------------------
# LOAD LLM
# --------------------------

@st.cache_resource
def load_llm():
    return ChatOllama(
        model="gemma2:2b",
        temperature=0
    )

# --------------------------
# PROCESS PDFS
# --------------------------

def process_pdfs(uploaded_files):

    docs = []

    for uploaded_file in uploaded_files:

        with tempfile.NamedTemporaryFile(
            delete=False,
            suffix=".pdf"
        ) as tmp:

            tmp.write(uploaded_file.read())
            temp_path = tmp.name

        loader = PyPDFLoader(temp_path)

        pdf_docs = loader.load()

        for doc in pdf_docs:
            doc.page_content = clean_text(
                doc.page_content
            )

        docs.extend(pdf_docs)

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=50
    )

    chunks = splitter.split_documents(docs)

    embeddings = load_embeddings()

    vectordb = Chroma.from_documents(
        documents=chunks,
        embedding=embeddings,
        persist_directory=CHROMA_DIR
    )

    return vectordb

# --------------------------
# RAG ANSWER
# --------------------------

def answer_question(question, vectordb):

    docs = vectordb.similarity_search(
        question,
        k=3
    )

    context = "\n\n".join(
        [doc.page_content for doc in docs]
    )

    prompt = f"""
You are a college document assistant.

Use ONLY the provided context.

If the answer cannot be found
in the context say:

'I could not find that information in the provided documents.'

Context:
{context}

Question:
{question}

Answer:
"""

    llm = load_llm()

    response = llm.invoke(prompt)

    sources = []

    for doc in docs:

        source = doc.metadata.get(
            "source",
            "Unknown"
        )

        if source not in sources:
            sources.append(source)

    return response.content, sources

# --------------------------
# STREAMLIT UI
# --------------------------

st.set_page_config(
    page_title="College Knowledge Assistant",
    page_icon="🎓",
    layout="wide"
)

st.title("🎓 College Knowledge Assistant")

st.write(
    "Upload college documents and ask questions."
)

uploaded_files = st.file_uploader(
    "Upload PDF Documents",
    type=["pdf"],
    accept_multiple_files=True
)

if uploaded_files:

    if st.button("Build Knowledge Base"):

        with st.spinner(
            "Processing documents..."
        ):

            vectordb = process_pdfs(
                uploaded_files
            )

            st.session_state["db"] = vectordb

        st.success(
            "Knowledge Base Created!"
        )

if "db" in st.session_state:

    question = st.text_input(
        "Ask a Question"
    )

    if question:

        with st.spinner(
            "Generating Answer..."
        ):

            answer, sources = answer_question(
                question,
                st.session_state["db"]
            )

        st.subheader("Answer")

        st.write(answer)

        st.subheader("Sources")

        for src in sources:
            st.write(f"📄 {src}")
```
